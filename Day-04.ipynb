```python
#Task : Which  Product Generated the highest revenue ?
#Calulate the revenue for each product
#Sort products by revenue in descending order
#Identify the top three revenue generating products
#Create a bar chart showimg revenue by product
#Display the revenue on the bars
#Create an interactive plotly bar chart with hover information showing Product name and Revenue
```

```python
#Task : Which  Product Generated the highest revenue ?
#Calulate the revenue for each product
product_revenue = (
    sales.groupby('Product')['Total_Amount_INR']
    .sum()
    .sort_values(ascending=False)
)
product_revenue.head(3)
```

```python
# Sort products by revenue in descending order
product_revenue = product_revenue.sort_values(ascending=False)
product_revenue.head()
```

```python
#Identify the top three revenue generating products
top_products = product_revenue.head(3)
top_products
```

```python
#Create a bar chart showing revenue by product

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.bar(product_revenue.index, product_revenue.values, color='skyblue')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.title('Product Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Display the revenue on the bars
#Create an interactive plotly bar chart with hover information showing Product name and Revenue

fig = px.bar(product_revenue, x=product_revenue.index, y=product_revenue.values, labels={'x': 'Product', 'y': 'Revenue'}, title='Product Revenue')
fig.show()
```

```python
#Task 2 : Which Cities Generated the Most Sales ?

#Calulate the total sales for each city
city_sales = sales.groupby('City')['Total_Amount_INR'].sum().sort_values(ascending=False)
city_sales
```

```python
#Rank cities from highest to lowest revenue
city_sales = city_sales.rank(ascending=False)
city_sales
```

```python
# Create a horizontal bar chart showing revenue by city
fig = px.bar(city_sales, x=city_sales.values, y=city_sales.index, orientation='h', labels={'x': 'Revenue', 'y': 'City'}, title='City Revenue')
fig.show()
```

```python
revenue_by_city = sales.groupby('City', as_index=False)['Total_Amount_INR'].sum()

fig = px.bar(
    revenue_by_city,
    x='City',
    y='Total_Amount_INR',
    title='Total Revenue by City',
    labels={
        'City': 'City',
        'Total_Amount_INR': 'Total Revenue (INR)'
    },
    hover_data={'Total_Amount_INR': ':,.2f'}
)

fig.update_layout(
    xaxis_title='City',
    yaxis_title='Total Revenue (INR)',
    xaxis_tickangle=-45
)

fig.show()
```
