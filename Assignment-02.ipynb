```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

tips = px.data.tips()      # restaurant bills and tips
iris = px.data.iris()      # flower measurements
stocks = px.data.stocks()  # weekly stock prices

tips.head()
```

```python
#Part A: Matplotlib
#Task 1: Line Chart

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12, 15, 14, 20, 24, 22]

plt.plot(months, sales, marker="o", color="black")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (thousands)")
plt.grid(True)
plt.show()
```

```python
#Task 2 : Bar Chart

import matplotlib.pyplot as plt

avg_tip = tips.groupby("day")["tip"].mean().reindex(["Thur", "Fri", "Sat", "Sun"])
plt.bar(avg_tip.index, avg_tip.values, color="white", edgecolor="black", hatch="///")
plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.show()
```

```python
#Task 3: Histogram and Scatter Plot

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(tips["total_bill"], bins=10, edgecolor="black", color="white")
axes[0].set_title("Distribution of Total Bill")
axes[0].set_xlabel("Total Bill")
axes[0].set_ylabel("Frequency")

axes[1].scatter(tips["total_bill"], tips["tip"], color="black")
axes[1].set_title("Total Bill vs Tip")
axes[1].set_xlabel("Total Bill")
axes[1].set_ylabel("Tip")

plt.tight_layout()
plt.savefig("tips_plots.png", dpi=150)
plt.show()
```

```python
#Part B: Seaborn
#Task 4: Count Plot and Bar Plot

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(data=tips, x="day", color="white", edgecolor="black")
plt.title("Number of Customers by Day")
plt.xlabel("Day")
plt.ylabel("Number of Customers")
plt.show()

sns.barplot(data=tips, x="day", y="total_bill", color="white", edgecolor="black")
plt.title("Average Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Bill")
plt.show()
```

```python
#Task 5: Box Plot

sns.boxplot(data=tips, x="day", y="total_bill", color="white")
plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()

sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Total Bill by Day and Sex")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
#Task 6: Scatter Plot with Colour and Style

sns.scatterplot(data=tips, x="total_bill", y="tip", hue="smoker", style="time")
plt.title("Total Bill and Tip by Smoking Status")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# Task 7: Pair Plot

sns.pairplot(
    iris,
    hue="species",
    vars=["sepal_length", "sepal_width", "petal_length", "petal_width"],
)
plt.show()
```

```python
# Part C: Plotly
# Task 8: Interactive Scatter Plot

import plotly.express as px

fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="day",
    size="size",
    hover_data=["sex", "time"],
    title="Bill vs Tip",
)
fig.show()
```

```python
# Task 9: Interactive Box Plot and Histogram

fig = px.box(
    tips, x="day", y="total_bill", color="sex", title="Total Bill by Day and Sex"
)
fig.show()

fig = px.histogram(
    tips,
    x="total_bill",
    color="time",
    nbins=15,
    title="Total Bill by Meal Time",
)
fig.show()
```

```python
# Task 10: Line Chart with Range Slider

fig = px.line(
    stocks,
    x="date",
    y=["GOOG", "AAPL", "AMZN"],
    title="Stock Prices (2018-2019)",
)
fig.update_xaxes(rangeslider_visible=True)
fig.show()
```
