```python
sales = pd.read_csv('sales_dataset_50_rows.csv')
sales
```

```python
sales.isnull().sum()
```

```python
sales['Quantity'].mean()
```

```python
sales['Quantity'].median()
```

```python
sales['Quantity'] = sales['Quantity'].fillna(2)
sales.isnull().sum()
```

```python
sales['Region'].unique()
```

```python
sales['Region'] = sales['Region'].replace('Noth', 'North' )
sales['Region'] = sales['Region'].replace('EAST', 'East' )
sales['Region'].unique()
```

```python
sales['Region'] = sales['Region'].fillna('North')
sales.isnull().sum()
```

```python
sales = sales.dropna()
sales.shape
```

```python
sales['Region'].value_counts()
```

```python
sales.duplicated().sum()
```

```python
sales = sales.drop_duplicates()
```

```python
sales['Unit_Price_INR'] = sales['Unit_Price_INR'].replace('three-forty-nine', '349')
sales['Unit_Price_INR'] = sales['Unit_Price_INR'].str.replace(',', '', regex=False).astype(float)
sales['Unit_Price_INR'].unique()
```

```python
sales.groupby('Region')['Unit_Price_INR'].sum()
```

```python
import matplotlib.pyplot as plt

year = [2015, 2016, 2017, 2018, 2019, 2020]
sales = [308, 200, 400, 700, 100, 500]

plt.plot(year, sales)
plt.xlabel("Year")
plt.ylabel("Sales")
plt.title("Sales Data")
plt.show()


```

```python
plt.barh(year, sales)
plt.xlabel("Sales")
plt.ylabel("Year")
plt.title("Sales Data")
plt.show()
```

```python
import seaborn as sns
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
#missing values in penguins dataset
penguins.isnull().sum()
```

```python
#distribution plot
sns.displot(data=penguins, x="flipper_length_mm")
plt.show()
```

```python
sns.catplot(data=penguins, x="species", y="flipper_length_mm", kind="box")
plt.show()
```

```python
sns.catplot(kind = 'strip', data=penguins, x="species", y="flipper_length_mm")
plt.show()
```

```python
sns.catplot(kind = 'swarm', data=penguins, x="species", y="flipper_length_mm", hue="sex")
plt.show()
```
