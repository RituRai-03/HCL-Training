```python
x=5
y=3
z=7
#comparison operators
print(x==y) #False
print(z==y) #False
print(x!=y) #True
print(x>y) #True
print(x<=z) #True
#logical operators
print(not(z>y)) #False
print(x>y and x<z) #True
print(x>y or x<z) #True
```
